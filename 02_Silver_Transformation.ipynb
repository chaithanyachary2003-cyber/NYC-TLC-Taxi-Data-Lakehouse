```python
df = spark.table("nyc_tlc.bronze.tlc_trips")
```

```python
bronze_df = spark.table("nyc_tlc.bronze.tlc_trips")

display(bronze_df.limit(10))
```

```python
from pyspark.sql.functions import col, sum, when

null_count = bronze_df.select([
    sum(
        when(col(c).isNull() | (col(c) == ""), 1)
        .otherwise(0)
    ).alias(c)
    for c in bronze_df.columns
])

display(null_count)
```

```python
bronze_df.filter(
    col('passenger_count').isNull() |
    (col('passenger_count') == "")
).select(
    "airport_fee",
    'congestion_surcharge',
    'passenger_count',
    'ratecodeid',
    'store_and_fwd_flag'
).show(20, truncate=False)
```

```python
total_rows = bronze_df.count()

distinct_rows = bronze_df.distinct().count()

print('Total rows:', total_rows)
print("Distinct rows:", distinct_rows)
print('Duplicate rows:', total_rows - distinct_rows)
```

```python
df_typed = bronze_df.select(
    col('dolocationid').cast('int'),
    col('extra').cast('double'),
    col('fare_amount').cast('double'),
    col('improvement_surcharge').cast('double'),
    col('mta_tax').cast('double'),
    col('payment_type').cast('int'),
    col('pulocationid').cast('int'),
    col('tip_amount').cast('double'),
    col('tolls_amount').cast('double'),
    col('total_amount').cast('double'),
    col('tpep_dropoff_datetime').cast('timestamp'),
    col('tpep_pickup_datetime').cast('timestamp'),
    col('trip_distance').cast('double'),
    col('vendorid').cast('int'),
    col('airport_fee').cast('double'),
    col('congestion_surcharge').cast('double'),
    col('passenger_count').cast('double').cast('int').alias("passenger_count"),
    col('ratecodeid').cast('double').cast('int').alias('ratecodeid'),
    col('store_and_fwd_flag').cast('string')
)

display(df_typed.limit(10))
```

```python
df_typed.printSchema()
```

```python
print('bronze rows:', bronze_df.count())
print('silver rows:', df_typed.count())
```

```python
null_count_silver = df_typed.select([
    sum(
        when(col(c).isNull(), 1)
        .otherwise(0)
    ).alias(c)
    for c in df_typed.columns
])

display(null_count_silver)
```

```python
print('negetive trip distance')
display(
    df_typed.filter(col('trip_distance')<0)
)

print('negetive fare ammount')
display(
    df_typed.filter(col('fare_amount')<0)
)

print('negetive total amount')
display(
    df_typed.filter(col("total_amount")<0)
)

print('drop off before pickup')
display(
    df_typed.filter(
        col('tpep_dropoff_datetime') < col('tpep_pickup_datetime')
    )
)
```

```python
silver_df = df_typed.filter(
    (col('fare_amount')>=0)&
    (col('total_amount')>=0)
)

print('bronze rows', bronze_df.count())
print('silver rows', silver_df.count())
print('rows removed', bronze_df.count()-silver_df.count())
```

```python
from pyspark.sql.functions import round, to_date, hour
silver_df = (silver_df.withColumn(
    "trip_duration_minutes",
    round(
        (
            col('tpep_dropoff_datetime').cast('long')
            - col('tpep_pickup_datetime').cast('long')
        )/60, 2
    )
)
.withColumn(
    "pickup_date",
    to_date(col('tpep_pickup_datetime'))
)
.withColumn(
    "pickup_hour",
    hour(col('tpep_pickup_datetime'))
)
)

display(
    silver_df.select(
        'tpep_pickup_datetime',
        'tpep_dropoff_datetime',
        'trip_duration_minutes',
        'pickup_date',
        'pickup_hour'
    ).limit(10)
)
```

```python
from pyspark.sql.functions import min, max, avg

silver_df.select(
    min('trip_duration_minutes').alias('min_duration'),
    max('trip_duration_minutes').alias('max_duration'),
    avg('trip_duration_minutes').alias('avg_duration')
).show()
```

```python
silver_df.write \
    .format('delta') \
        .mode('overwrite') \
            .saveAsTable('nyc_tlc.silver.cleaned_tlc_trips')
```

```python
spark.sql("""
          select count(*) as silver_row_count
          from nyc_tlc.silver.cleaned_tlc_trips""").show()
```

```python
spark.sql('''
          select * from nyc_tlc.silver.cleaned_tlc_trips
          limit 10''').show()
```

```python
from pyspark.sql.functions import count

daily_taxi_performance = (
    silver_df
    .groupBy('pickup_date')
    .agg(
        count("*").alias('trip_count'),
        round(sum('total_amount'),2).alias('total_revenue'),
        round(avg('fare_amount'), 2).alias('average_fare'),
        round(avg('trip_distance'),2).alias('average_trip_distance')
    )
    .orderBy('pickup_date')
)

display(daily_taxi_performance)
```

```python
daily_taxi_performance.write \
    .format('delta') \
        .mode('overwrite') \
            .saveAsTable('nyc_tlc.gold.daily_taxi_performance')
```

```python
spark.sql('''
          select * from nyc_tlc.gold.daily_taxi_performance
          limit 10''').show()
```

```python
payment_analysis = (
    silver_df
    .groupBy('payment_type')
    .agg(
        count("*").alias('total_count'),
        round(sum('total_amount'), 2).alias('trip_revenue'),
        round(avg('fare_amount'), 2).alias('average_fare')
    )
    .orderBy('payment_type')
)

display(payment_analysis)
```

```python
payment_analysis.write \
    .format('delta') \
    .mode('overwrite') \
    .saveAsTable('nyc_tlc.gold.payment_analysis')
```

```python
spark.sql('''
          select * from nyc_tlc.gold.payment_analysis
          ''').show()
```

```python
peak_hour_analysis =(
    silver_df
    .groupBy('pickup_hour')
    .agg(
        count('*').alias('trip_count'),
        round(avg('fare_amount'),2).alias('average_fare'),
        round(avg('trip_duration_minutes'),2).alias('avg_trip_duration')
    )
    .orderBy('pickup_hour')
)

display(peak_hour_analysis)
```

```python
peak_hour_analysis.write \
    .format('delta') \
        .mode('overwrite') \
            .saveAsTable('nyc_tlc.gold.peak_hour_analysis')
```

```python
spark.table('nyc_tlc.gold.peak_hour_analysis').show()
```
